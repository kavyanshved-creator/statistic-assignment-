```python
"""2.
Given the data: Group A (users shown a new Zomato home page) had an 
average order value of ₹350 from 30 users, and Group B (old home page)
 had an average order value of ₹320 from 30 users. Assume both groups have
a standard deviation of ₹50. Calculate the p-value using a two-sample 
t-test (you may use Python, R, or an online calculator), and decide whether
to accept or reject the null hypothesis at a 0.05 significance level.
Use scipy.stats.ttest_ind in Python or any online t-test calculator if
 you are not comfortable with code.</em>"""


from scipy.stats import ttest_ind_from_stats

# Given data
mean_a = 350
mean_b = 320
std_a = 50
std_b = 50
n_a = 30
n_b = 30

# Perform two-sample t-test
t_stat, p_value = ttest_ind_from_stats(
    mean1=mean_a, std1=std_a, nobs1=n_a,
    mean2=mean_b, std2=std_b, nobs2=n_b,
    equal_var=True
)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```
