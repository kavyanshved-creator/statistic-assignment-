```python
"""
1.
Generate 100 random numbers in Python that follow a normal distribution with
 mean 50 and standard deviation 10 using numpy, then plot the bell curve using matplotlib.
"""

import numpy as np
import matplotlib.pyplot as plt

mean = 50
standard_deviation = 10
numbers = np.random.default_rng().normal(
    loc=mean,
    scale=standard_deviation,
    size=100,
)
print(numbers)

x = np.linspace(mean - 4 * standard_deviation, mean + 4 * standard_deviation, 300)
pdf = (1 / (standard_deviation * np.sqrt(2 * np.pi))) * np.exp(
    -0.5 * ((x - mean) / standard_deviation) ** 2
)

plt.hist(numbers, bins=15, density=True, alpha=0.6, edgecolor="black", label="Generated numbers")
plt.plot(x, pdf, color="red", linewidth=2, label="Normal distribution (mean=50, SD=10)")
plt.title("Normal Distribution: Mean 50, Standard Deviation 10")
plt.xlabel("Value")
plt.ylabel("Probability density")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

```python
"""
2.
Given an array of 20 daily step counts from your fitness tracker app, calculate 
the mean, median, and mode, and check if they are approximately equal as expected 
in a normal distribution.<br><br><em><strong>Hint:</strong> Use numpy.mean(),
numpy.median(), and scipy.stats.mode().</em>
 """

import numpy as np
from scipy import stats

step_counts = np.array([
    6500, 9500, 7000, 9000, 7200, 8800, 7500, 8500, 7600, 8400,
    7700, 8300, 7800, 8200, 7900, 8100, 8000, 8000, 8000, 8000,
])

mean_steps = np.mean(step_counts)
median_steps = np.median(step_counts)
mode_steps = stats.mode(step_counts, keepdims=False).mode.item()

print(f"Mean: {mean_steps:.0f} steps")
print(f"Median: {median_steps:.0f} steps")
print(f"Mode: {mode_steps:.0f} steps")

tolerance = 200
approximately_equal = (
    abs(mean_steps - median_steps) <= tolerance
    and abs(median_steps - mode_steps) <= tolerance
)
print(f"Approximately equal (within {tolerance} steps): {approximately_equal}")
```

```python
"""
3.
Simulate the scores of 200 players in an online game (e.g., PUBG, Free Fire)
 using a normal distribution with mean 70 and standard deviation 15, then count
   how many players scored within one standard deviation (between 55 and 85) of 
   the mean.<br><br><em><strong>Hint:</strong> Use the 68-95-99 rule to check your answer.</em>
"""

import numpy as np

mean_score = 70
standard_deviation = 15
player_scores = np.random.default_rng().normal(
    loc=mean_score,
    scale=standard_deviation,
    size=200,
)

lower_bound = mean_score - standard_deviation  # 55
upper_bound = mean_score + standard_deviation  # 85
within_one_std = np.count_nonzero(
    (player_scores >= lower_bound) & (player_scores <= upper_bound)
)

print(f"Scores within one standard deviation ({lower_bound:.0f}-{upper_bound:.0f}): {within_one_std} out of {player_scores.size}")
print(f"Percentage: {within_one_std / player_scores.size:.1%}")
print("The 68-95-99.7 rule predicts about 68% (approximately 136 players) within one standard deviation.")
```
