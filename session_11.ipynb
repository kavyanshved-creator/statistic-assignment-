```python
"""
1.
Create a Python list to represent a playlist of 5 song durations in seconds 
(e.g., [210, 180, 245, 200, 175]). Print the total duration of the playlist."""

playlist = [210, 180, 245, 200, 175]
total_duration = sum(playlist)
print("Total duration of the playlist:", total_duration, "seconds")
```

```python
"""
2.
Represent a 3x3 matrix in Python to store the ratings (1-5) given by 3 friends
 to 3 different food items on Swiggy. Print the matrix and display the rating
each friend gave to the second food item.
"""

import numpy as np

# Ratings given by 3 friends to 3 food items on Swiggy
ratings = np.array([
    [4, 5, 3],
    [3, 4, 5],
    [5, 2, 4]
])

# Print the 3x3 matrix
print("Food Ratings Matrix:")
print(ratings)

# Display ratings given by each friend to the second food item
print("\nRatings for the second food item:")
print(ratings[:, 1])
```

```python
"""3.
Given a matrix representing the number of orders for 4 different dishes across 
3 days (rows: days, columns: dishes), write code to calculate and print the total 
orders for each dish."""

import numpy as np

# Rows represent days and columns represent dishes
orders = np.array([
    [10, 15, 20, 12],
    [12, 18, 15, 10],
    [8,  14, 22, 16]
])

# Calculate total orders for each dish
total_orders = np.sum(orders, axis=0)

print("Total orders for each dish:")
print(total_orders)
```

```python
"""
4.
Imagine you have a table of 5 Flipkart products with columns for price and quantity 
sold. Represent this data as a matrix in Python, then write a function to calculate 
the total revenue for each product.<br><br><em><strong>Hint:</strong> Multiply price 
and quantity for each row, then print the results.</em>"""

import numpy as np

# Matrix columns: Price, Quantity Sold
products = np.array([
    [500, 10],
    [800, 5],
    [300, 12],
    [1000, 4],
    [200, 15]
])

# Function to calculate total revenue for each product
def calculate_revenue(products):
    revenue = products[:, 0] * products[:, 1]
    print("Total revenue for each product:")
    print(revenue)

# Call the function
calculate_revenue(products)
```
