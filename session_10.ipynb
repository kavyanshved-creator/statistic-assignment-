```python
"""
1.
Write a Python function bayes_posterior(prior, likelihood, evidence) that calculates
 and returns the posterior probability using Bayes theorem formula for given values."""

def bayes_posterior(prior, likelihood, evidence):
    """Return P(H|E) = P(E|H) * P(H) / P(E)."""
    if not all(0 <= probability <= 1 for probability in (prior, likelihood, evidence)):
        raise ValueError("prior, likelihood, and evidence must be between 0 and 1")
    if evidence == 0:
        raise ValueError("evidence must be greater than 0")

    return (prior * likelihood) / evidence


# Example: prior P(H)=0.1, likelihood P(E|H)=0.8, evidence P(E)=0.26
posterior = bayes_posterior(prior=0.1, likelihood=0.8, evidence=0.26)
print(f"Posterior probability: {posterior:.4f}")
```

```python
"""2.
Given the following scenario: On Swiggy, 10% of restaurants are 'premium', and 80% 
of premium restaurants offer a discount, while only 20% of regular restaurants do. 
If you see a discount, use Bayes theorem in code to find the probability that the 
restaurant is premium.<br><br><em><strong>Hint:</strong> Define prior, likelihood, 
and evidence in your code with comments.</em>"""

# Prior probability
prior_premium = 0.10  # 10% of restaurants are premium

# Likelihood
likelihood_discount_premium = 0.80  # 80% of premium restaurants offer discounts
likelihood_discount_regular = 0.20  # 20% of regular restaurants offer discounts

# Evidence: Probability that a restaurant offers a discount
evidence = (likelihood_discount_premium * prior_premium) + (likelihood_discount_regular * (1 - prior_premium))

# Bayes' theorem: Probability that a restaurant is premium given that it offers a discount
probability = (likelihood_discount_premium * prior_premium) / evidence

print("Probability that the restaurant is premium:", probability * 100, "%")
```

```python
"""
3.
Simulate a basic spam detection filter: Given a list of 20 email subjects 
(some spam, some not), and the probability that 'free' appears in spam vs non-spam emails,
 write code to calculate the probability an email is spam if it contains 'free', using Bayes theorem."""

# List of 20 email subjects
emails = [
    "Get free gifts now", "Meeting scheduled for tomorrow",
    "Claim your free prize", "Project report submission",
    "Free entry in lucky draw", "Your order has been shipped",
    "Win free cash today", "Team meeting at 10 AM",
    "Get free coupons", "Your bank statement",
    "Free vacation offer", "Assignment submission reminder",
    "Congratulations, you won free money", "Interview scheduled",
    "Free shopping vouchers", "Weekly progress report",
    "Claim free rewards now", "Your password has been changed",
    "Free bonus offer", "College exam timetable"
]

# Prior probabilities
prior_spam = 10 / 20       # 10 spam emails out of 20
prior_non_spam = 10 / 20   # 10 non-spam emails out of 20

# Likelihoods
# Assume 'free' appears in 8 spam emails and 1 non-spam email
likelihood_free_spam = 8 / 10
likelihood_free_non_spam = 1 / 10

# Evidence: Probability that 'free' appears in an email
evidence = (likelihood_free_spam * prior_spam) + \
           (likelihood_free_non_spam * prior_non_spam)

# Bayes' theorem
probability_spam = (likelihood_free_spam * prior_spam) / evidence

print("Probability that an email containing 'free' is spam:",
      round(probability_spam * 100, 2), "%")
```

```python
"""
4.
Use ChatGPT or Copilot to generate a real-world Bayes theorem example related to 
any app you use (Instagram, Zomato, Flipkart, etc.), then write code to solve it 
and submit both the prompt you used and your code."""

"""
Generate a real-world example of Bayes' theorem related to Instagram.
Suppose 30% of Instagram users click on an advertisement.
Among users who purchase a product, 70% clicked the advertisement.
Among users who do not purchase, 10% clicked the advertisement.
Write Python code to calculate the probability that a user purchases
the product given that they clicked the advertisement.
"""

# Bayes' Theorem Example: Instagram Advertisement

# Prior probabilities
prior_purchase = 0.30       # 30% users purchase the product
prior_no_purchase = 1 - prior_purchase

# Likelihood probabilities
likelihood_click_purchase = 0.70
likelihood_click_no_purchase = 0.10

# Evidence: Probability that a user clicks the advertisement
evidence = (likelihood_click_purchase * prior_purchase) + \
           (likelihood_click_no_purchase * prior_no_purchase)

# Bayes' theorem
probability = (likelihood_click_purchase * prior_purchase) / evidence

print("Probability that a user purchases after clicking the ad:",
      round(probability * 100, 2), "%")
```
