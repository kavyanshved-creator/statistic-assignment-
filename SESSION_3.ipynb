```python
"""
1.
Given an array of daily step counts from your fitness app for one week 
(e.g., [4500, 7000, 5000, 8000, 4000, 9000, 6000]), calculate the variance
 by hand using the variance formula and show each calculation step.
 """

steps = [4500, 7000, 5000, 8000, 4000, 9000, 6000]

# Treat the 7 recorded days as the full population for this week.
mean_steps = sum(steps) / len(steps)
squared_deviations = [(count - mean_steps) ** 2 for count in steps]
variance = sum(squared_deviations) / len(steps)

print("Population variance formula: variance = sum((x - mean)^2) / 7")
print(f"Mean = (4500 + 7000 + 5000 + 8000 + 4000 + 9000 + 6000) / 7 = {mean_steps:.2f}\n")
print("Steps   Deviation from mean   Squared deviation")
for count, squared_deviation in zip(steps, squared_deviations):
    print(f"{count:>5}   {count - mean_steps:>+18.2f}   {squared_deviation:>18,.2f}")

print(f"\nSum of squared deviations = {sum(squared_deviations):,.2f}")
print(f"Variance = {sum(squared_deviations):,.2f} / 7 = {variance:,.2f} steps^2")
```

```python
"""2.
Write a Python function called calculate_standard_deviation(scores) that 
takes a list of exam scores and returns the standard deviation rounded to 
two decimal places."""


def calculate_standard_deviation(scores):
    if not scores:
        return 0
    mean = sum(scores) / len(scores)
    squared_deviations = [(x - mean) ** 2 for x in scores]
    variance = sum(squared_deviations) / len(scores)
    standard_deviation = variance ** 0.5
    return round(standard_deviation, 2)



scores = [85, 90, 78, 92, 88]
calculate_standard_deviation(scores)
```

```python
"""
3.
Imagine two friends track their daily spending on Zomato over 5 days.
 Friend A spends [200, 200, 200, 200, 200] and Friend B spends 
 [100, 300, 150, 400, 50]. Calculate the standard deviation for both 
 and explain in 2-3 lines who has more consistent spending and why.
"""

friend_A = [200, 200, 200, 200, 200]
friend_B = [100, 300, 150, 400, 50]

print("Friend A's standard deviation:", calculate_standard_deviation(friend_A))
print("Friend B's standard deviation:", calculate_standard_deviation(friend_B))
print("Friend A has more consistent spending because the standard deviation is 0.00, meaning the spending is the same every day.")
print("Friend B's standard deviation is 130.38, showing much greater day-to-day variation in spending.")
```

```python
"""
4.
You are given the monthly salary data of 6 employees in a startup: 
[28000, 29000, 27000, 60000, 26500, 27500]. Calculate the variance 
and standard deviation, then interpret in 2-3 lines what the high or
 low spread means for this team."""

salaries = [28000, 29000, 27000, 60000, 26500, 27500]
mean_salary = sum(salaries) / len(salaries)
squared_deviations = [(salary - mean_salary) ** 2 for salary in salaries]
variance = sum(squared_deviations) / len(salaries)
standard_deviation = variance ** 0.5

print("Population variance formula: variance = sum((x - mean)^2) / 6")
print(f"Mean salary = {sum(salaries):,} / {len(salaries)} = {mean_salary:,.2f}\n")
print("Salary   Deviation from mean   Squared deviation")
for salary, squared_deviation in zip(salaries, squared_deviations):
    print(f"{salary:>6}   {salary - mean_salary:>+17,.2f}   {squared_deviation:>18,.2f}")

print(f"\nSum of squared deviations = {sum(squared_deviations):,.2f}")
print(f"Variance = {sum(squared_deviations):,.2f} / {len(salaries)} = {variance:,.2f} salary^2")
print(f"Standard deviation = sqrt({variance:,.2f}) = {standard_deviation:,.2f} salary units\n")
print("The spread is high: five salaries cluster between 26,500 and 29,000, while one salary is 60,000.")
print("This standard deviation shows that salaries vary substantially, and the 60,000 salary pulls the mean upward.")
```
