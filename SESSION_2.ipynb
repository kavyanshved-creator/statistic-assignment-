```python
"""
3.
Write a function most_common_rating(ratings) that takes a list of Flipkart
 product ratings (e.g., [5, 4, 4, 3, 5, 4, 2, 4]) and returns the mode 
 (most frequent rating)."""

from statistics import mode

def most_common_rating(ratings):
    return mode(ratings)

ratings = [5, 4, 4, 3, 5, 4, 2, 4]

result = most_common_rating(ratings)

print("Most common rating:", result)
```

```python
"""4.
Given three lists representing YouTube video views for three different
channels, compare the mean, median, and mode for each channel, and 
decide which channel's data is most affected by
outliers.
Try using one list with a very high value (e.g.,
[100, 120, 110, 105, 5000]) to see the effect
 on mean vs median.</em>"""

import statistics

channel1 = [100, 120, 110, 105, 5000]
channel2 = [200, 220, 210, 230, 240]
channel3 = [100, 100, 110, 110, 120]

channels = {
    "Channel 1": channel1,
    "Channel 2": channel2,
    "Channel 3": channel3
}

for name, views in channels.items():
    mean = statistics.mean(views)
    median = statistics.median(views)
    mode = statistics.multimode(views)

    print(name)
    print("Mean:", mean)
    print("Median:", median)
    print("Mode:", mode)
    print()
```
