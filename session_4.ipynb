```python
"""
1.
Given the daily order counts for a Swiggy delivery partner over 10 days: 
[12, 15, 17, 14, 13, 16, 200, 18, 14, 15], plot a simple graph (hand-drawn 
or using any tool) and describe whether the data is left skewed, right
 skewed, or symmetrical. Explain your reasoning based on the shape.
 """

import matplotlib.pyplot as plt

orders = [12, 15, 17, 14, 13, 16, 200, 18, 14, 15]
days = list(range(1, len(orders) + 1))

plt.figure(figsize=(8, 4))
plt.bar(days, orders, color="mediumseagreen")
plt.xticks(days)
plt.xlabel("Day")
plt.ylabel("Number of orders")
plt.title("Daily Orders for a Swiggy Delivery Partner")
plt.tight_layout()
plt.show()

print(
    "The distribution is right-skewed: most days have between 12 and 18 orders, "
    "while the unusually high count of 200 creates a long tail toward the right."
)
```

```python
"""
2.
Take the following list of Instagram influencer follower counts (in thousands):
 [5, 7, 8, 8, 9, 10, 12, 15, 95]. Calculate the mean and median, then state 
 which is higher and what that tells you about the skewness of the 
 data.<br><br><em><strong>Hint:</strong> If the mean is pulled away
   from the median, the data is skewed in that direction.</em>
   """

import statistics

followers = [5, 7, 8, 8, 9, 10, 12, 15, 95]
mean_followers = statistics.mean(followers)
median_followers = statistics.median(followers)

print(f"Mean: {mean_followers:.2f} thousand followers")
print(f"Median: {median_followers:.2f} thousand followers")
print(
    "The mean (19.89 thousand) is higher than the median (9 thousand), "
    "showing that the data is right-skewed. The unusually high value of 95 "
    "pulls the mean upward and creates a tail toward larger follower counts."
)
```

```python
"""
3.
Find a real-world example of right-skewed data from any app you use 
(like Zomato order amounts, Paytm wallet balances, or YouTube video views).
 Briefly describe the dataset and explain why you think it is right skewed.
 """


video_views = [120, 250, 300, 450, 600, 800, 1200, 2500, 50000]
print("Example: YouTube video views")
print("Sample view counts:", video_views)
print(
    "Most videos in this sample have relatively few views, while one video has "
    "50,000 views. This very large value stretches the distribution toward higher "
    "view counts, creating a long right tail; therefore, the data is right-skewed."
)
```

```python
"""
4.
Given this dataset of Flipkart product ratings: [3, 3, 4, 4, 4, 5, 5, 5, 5, 5],
 draw a rough graph and state if the distribution is symmetrical, left skewed, 
 or right skewed. Justify your answer in one line.
"""

import matplotlib.pyplot as plt
from collections import Counter

ratings = [3, 3, 4, 4, 4, 5, 5, 5, 5, 5]
rating_counts = Counter(ratings)

plt.bar(rating_counts.keys(), rating_counts.values(), color="cornflowerblue")
plt.xticks([3, 4, 5])
plt.xlabel("Product rating")
plt.ylabel("Number of ratings")
plt.title("Flipkart Product Ratings")
plt.show()

print("Ratings frequency:")
for rating in [3, 4, 5]:
    print(f"{rating}: {rating_counts[rating]}")
print(
    "The distribution is left-skewed: most ratings are 5, and the smaller "
    "ratings (4 and 3) form a tail extending toward the left."
)
```
