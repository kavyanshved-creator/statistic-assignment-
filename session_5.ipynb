```python
"""
1.
List 3 examples of 'population' and 'sample' in the context of apps you use daily 
(like Instagram followers, Zomato restaurants, or Flipkart products). For each, 
clearly define what would be considered the population and what would be a sample.

ANS)

1. Examples of population and sample

A population is the complete group we want to study. A sample is a smaller
 part of that group selected for observation.

1. Instagram followers
Population: All followers of a particular Instagram account.
Sample: 100 followers randomly selected from that account to answer a survey
about how often they use Instagram.

2. Zomato restaurants 
Population: All restaurants listed on Zomato in Mumbai.
Sample: 50 restaurants randomly selected from the Mumbai listings to study customer ratings.

3. Flipkart products
Population: All smartphone products listed on Flipkart.
Sample: 200 smartphones randomly selected from those listings to compare prices and ratings."""
```

```python
"""
2.
Write a Python script that simulates random sampling: given a list of 1000 fake 
user IDs (e.g., user1 to user1000), select 50 random users to simulate a survey 
sample. Print the selected user IDs.
Use the random.sample() function from Python's random module.
"""

import random

user_ids = [f"user_ids {i}" for i in range(1, 1001)]
survey_sample = random.sample(user_ids, 50)

print("Selected 50 users for the survey:")
for user_id in survey_sample:
    print(user_id)
```

```python
"""
4.
Given this scenario: A survey about music preferences is only sent to users who 
recently streamed Bollywood songs on Spotify. Identify and explain the type of 
bias this introduces in the sample."""

"""**Answer: 

Selection Bias**
This introduces **selection bias** because the survey is sent only to users who
 recently streamed Bollywood songs on Spotify. Their music preferences may not
   represent all Spotify users, including those who prefer English, Punjabi, or other genres.

As a result, the survey may produce misleading conclusions about the overall 
music preferences of Spotify users."""
```
