```python
"""
1.
Create two Python lists: one representing the number of hours you spend on Instagram
 each day for a week, and another for the number of posts you make each day. Plot these 
 as a scatter plot using matplotlib and visually describe whether the relationship looks
   like a positive, negative, or zero correlation."""

import matplotlib.pyplot as plt

instagram_hours = [1, 2, 2.5, 3, 4, 5, 6] # Number of hours spent on Instagram each day for a week


posts_made = [2, 3, 4, 5, 6, 8, 10] # Number of posts made each day


plt.scatter(instagram_hours, posts_made, color='blue')
plt.title('Instagram Hours vs. Posts Made')
plt.xlabel('Hours spent on Instagram per day')
plt.ylabel('Posts made per day')
plt.grid(True)
plt.show()
```
